3.2.1 Limits And Continuity

```python
import sympy as sp

# Define the variable and the function
x = sp.symbols('x')
f = (x**2 - 1)/(x - 1)

# Calculate the limit as x approaches 1
limit_at_1 = sp.limit(f, x, 1)
print(f"Limit as x approaches 1: {limit_at_1}")

# Check continuity: f(1) exists and equals the limit at 1
f_at_1 = f.subs(x, 1)
if f_at_1 == limit_at_1:
    print("The function is continuous at x = 1")
else:
    print("The function is not continuous at x = 1")
```

3.2.2 Derivatives

```python
import sympy as sp

# Define the variable and the function
x = sp.symbols('x')
f = sp.sin(x) * sp.exp(x)

# Compute the first derivative
derivative = sp.diff(f, x)
print(f"The derivative of f(x) = sin(x) * exp(x) is: {derivative}")

# Compute the second derivative
second_derivative = sp.diff(derivative, x)
print(f"The second derivative is: {second_derivative}")
```

3.2.3 Taylor's Theorem

```python
import sympy as sp

# Define the variable and the function
x = sp.symbols('x')
f = sp.sin(x)

# Compute the Taylor series expansion of sin(x) around x=0 (Maclaurin series) up to order 5
taylor_series = f.series(x, 0, 6)
print(f"Taylor series of sin(x) around x=0 up to order 5: {taylor_series}")
```
